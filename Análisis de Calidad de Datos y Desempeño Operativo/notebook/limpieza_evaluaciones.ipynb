#### Tareas realizadas :
Dado que la columna comentarios presenta valores nulos e inconsistencias en su redacción, se decide utilizar la variable score_desempeño como referencia principal para el análisis, ya que representa una medición objetiva del desempeño del empleado.

Los comentarios se consideran un campo complementario y opcional, por lo que se procede a estandarizarlos en función del puntaje registrado. De esta manera se logra coherencia entre ambas variables y se facilita el análisis posterior.

### Conclusión : 
Se realiza un chequeo general de la tabla, observando que cuenta con 9 filas y 3 columnas. Durante la revisión se detecta que la variable score_desempeño presenta un tipo de dato incorrecto, encontrándose como object cuando debería ser una variable numérica entera, por lo que se procede a su conversión.

Asimismo, se identifican valores repetidos en las columnas score_desempeño y comentarios, situación esperable dado que una representa una variable cuantitativa discreta y la otra una variable categórica. También se observan valores nulos únicamente en la columna comentarios, lo cual no afecta el análisis ya que se trata de un campo complementario. Por último, se detectan registros donde aparecen valores descriptivos en score_desempeño que corresponden en realidad a la columna comentarios, por lo que se corrigen para mantener coherencia en los tipos de datos.

#### Tareas realizadas en columna 'score_desempeño'
Se detectan valores de tipo string en la variable score_desempeño, los cuales no corresponden a la escala numérica definida para el análisis. En estos casos, se procede a su conversión al valor numérico equivalente, tomando “excelente” como 10, según los criterios establecidos por la empresa.
Se procede a la conversión de la variable 'score_desempeño' de object a numero entero

Asimismo, la empresa proporcionó una escala de equivalencias entre puntaje y comentarios cualitativos, la cual se utilizará para completar y estandarizar la columna comentarios, garantizando coherencia entre ambas variables.

| Desempeño | comentarios|
| --------- | ---------- |
|    10     |  excelente |
|  9 y 8    |  muy bueno |
|    7      |buen desempeño|
|    6      |  aceptable     |
|  4 y 5    |  regular   |

#### Tratamiento columna 'desempeño'

```python
df_evaluaciones.loc[
    df_evaluaciones['score_desempeño'] == '10'
    ,'comentarios'
] = 'excelente'
```

```python
df_evaluaciones.loc[
    df_evaluaciones['score_desempeño'] == '8'
    ,'comentarios'
] = 'muy bueno'
```

```python
df_evaluaciones.loc[
    df_evaluaciones['score_desempeño'] == '6'
    ,'comentarios'

] = 'aceptable'
```

```python
df_evaluaciones
```

```python
df_evaluaciones = df_evaluaciones[df_evaluaciones['id_empleado'] != 169 ]
```

```python
df_evaluaciones = df_evaluaciones[df_evaluaciones['id_empleado'] != 152]
```

#### Tareas realizadas en columna 'comentarios' :
Se completaron y corrigieron los valores faltantes y existentes en la columna comentarios, tomando como referencia los puntajes de la columna score_desempeño y la tabla de equivalencias proporcionada por la empresa.

El registro de id_empleado 169 (ascenso de Ventas a RRHH) presentaba ausencia total de datos en score_desempeño y comentarios, por lo que se decidió eliminarlo del análisis de desempeño debido a la falta de información objetiva.

Por otro lado, id_empleado 152 corresponde a otra empleada con el mismo nombre pero distinta edad y sector (Ventas). Si bien el registro se mantiene en el dataset original por tratarse de una persona distinta, también se excluye del análisis de desempeño al no contar con métricas evaluativas.
